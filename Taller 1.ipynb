##  Ejercicios SQL sobre el esquema HR en Oracle

A continuación se presenta la solución de los ejercicios 1 al 15 consultando directamente la base de datos Oracle Xe.

### 1. Proyectar el nombre, apellido y salario de todos los empleados de la base de datos.

```python
query = "SELECT first_name, last_name, salary FROM employees"
df1 = pd.read_sql_query(query, engine_oracle)
df1
```

### 2. Proyectar el código y nombre de los departamentos.

```python
query = "SELECT department_id, department_name FROM departments"
df2 = pd.read_sql_query(query, engine_oracle)
df2
```

### 3. Seleccionar los empleados cuyo salario sea superior a 10.000 dólares.

```python
query = "SELECT * FROM employees WHERE salary > 10000"
df3 = pd.read_sql_query(query, engine_oracle)
df3
```

### 4. Ordenar los empleados por salario. Mostrar de forma ascendente y descendente.

```python
df_desc = pd.read_sql_query("SELECT * FROM employees ORDER BY salary DESC", engine_oracle)
df_asc = pd.read_sql_query("SELECT * FROM employees ORDER BY salary ASC", engine_oracle)

print("--- ORDENADO DESCENDENTE ---")
display(df_desc.head(5))
print("\n--- ORDENADO ASCENDENTE ---")
display(df_asc.head(5))
```

### 5. Mostrar los empleados cuyo nombre inicien por la letra L.

```python
query = "SELECT * FROM employees WHERE LOWER(first_name) LIKE 'l%'"
df5 = pd.read_sql_query(query, engine_oracle)
df5
```

### 6. Consultar la cantidad de empleados de la Base de Datos.

```python
query = "SELECT COUNT(*) AS total_empleados FROM employees"
df6 = pd.read_sql_query(query, engine_oracle)
df6
```

### 7. Consultar la cantidad de empleados por cada uno de los departamentos.

```python
query = """
SELECT e.department_id, d.department_name, COUNT(*) AS qty_employees 
FROM employees e 
LEFT JOIN departments d ON e.department_id = d.department_id 
GROUP BY e.department_id, d.department_name 
ORDER BY e.department_id
"""
df7 = pd.read_sql_query(query, engine_oracle)
df7
```

### 8. ¿Cuál es el empleado con el mayor salario de la organización?

```python
query = "SELECT * FROM employees WHERE salary = (SELECT MAX(salary) FROM employees)"
df8 = pd.read_sql_query(query, engine_oracle)
df8
```

### 9. ¿Cuál es el promedio del salario de los empleados?

```python
query = "SELECT ROUND(AVG(salary), 2) AS promedio_salario FROM employees"
df9 = pd.read_sql_query(query, engine_oracle)
df9
```

### 10. ¿Cuál es el promedio del salario de los empleados por departamento?

```python
query = """
SELECT e.department_id, d.department_name, ROUND(AVG(e.salary), 2) AS avg_salary 
FROM employees e 
LEFT JOIN departments d ON e.department_id = d.department_id 
GROUP BY e.department_id, d.department_name 
ORDER BY e.department_id
"""
df10 = pd.read_sql_query(query, engine_oracle)
df10
```

### 11. Cantidad de empleados que iniciaron a trabajar en la organización por cada uno de los meses del año.

```python
query = """
SELECT EXTRACT(MONTH FROM hire_date) AS mes, COUNT(*) AS cantidad_empleados 
FROM employees 
GROUP BY EXTRACT(MONTH FROM hire_date) 
ORDER BY mes
"""
df11 = pd.read_sql_query(query, engine_oracle)
df11
```

### 15. Empleados donde el nombre contiene vocal "i" o "u" y también la consonante "s". Utilizando expresiones regulares.

```python
query = "SELECT employee_id, first_name, last_name FROM employees WHERE REGEXP_LIKE(first_name, '([iu].*s|s.*[iu])', 'i')"
df15 = pd.read_sql_query(query, engine_oracle)
df15
```

### 16. Automatizar en una lista de cadenas cada una de las consultas anteriores, como arreglo de consultas en Pandas

```python
queries = [
    "SELECT first_name, last_name, salary FROM employees",  # 1
    "SELECT department_id, department_name FROM departments",  # 2
    "SELECT * FROM employees WHERE salary > 10000",  # 3
    "SELECT * FROM employees ORDER BY salary DESC",  # 4 (desc)
    "SELECT * FROM employees ORDER BY salary ASC",  # 4 (asc)
    "SELECT * FROM employees WHERE LOWER(first_name) LIKE 'l%'",  # 5 (L...)
    "SELECT COUNT(*) AS total_empleados FROM employees",  # 6
    "SELECT e.department_id, d.department_name, COUNT(*) AS qty_employees FROM employees e LEFT JOIN departments d ON e.department_id = d.department_id GROUP BY e.department_id, d.department_name ORDER BY e.department_id",  # 7
    "SELECT * FROM employees WHERE salary = (SELECT MAX(salary) FROM employees)",  # 8
    "SELECT ROUND(AVG(salary), 2) AS promedio_salario FROM employees",  # 9
    "SELECT e.department_id, d.department_name, ROUND(AVG(e.salary), 2) AS avg_salary FROM employees e LEFT JOIN departments d ON e.department_id = d.department_id GROUP BY e.department_id, d.department_name ORDER BY e.department_id",  # 10
    "SELECT EXTRACT(MONTH FROM hire_date) AS mes, COUNT(*) AS cantidad_empleados FROM employees GROUP BY EXTRACT(MONTH FROM hire_date) ORDER BY mes",  # 11
    "SELECT * FROM employees WHERE LOWER(first_name) LIKE 'st%'",  # 12
    "SELECT * FROM employees WHERE LOWER(first_name) LIKE '%s'",  # 13
    "SELECT * FROM employees WHERE LOWER(first_name) LIKE '%mar%'",  # 14
    "SELECT employee_id, first_name, last_name FROM employees WHERE REGEXP_LIKE(first_name, '([iu].*s|s.*[iu])', 'i')"  # 15
]

print(" Ejecutando automatización de consultas con Pandas...")
for idx, q in enumerate(queries, 1):
    df_result = pd.read_sql_query(q, engine_oracle)
    print(f"\n==========================================================")
    print(f" CONSULTA {idx}: {q[:75]}...")
    print(f"==========================================================")
    display(df_result.head(3))
```
